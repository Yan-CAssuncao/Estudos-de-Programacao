````python
cargo = int(input("Digite o cargo do funcionario: "))
sal_atual = float(input("Digite o salário atual: "))

if cargo == 1:
    reaj = (sal_atual * 7)/100
elif cargo == 2:
    reaj = (sal_atual * 9)/100
elif cargo == 3:
    reaj = (sal_atual * 5)/100
else:
    reaj = (sal_atual * 12)/100

print(f"O resjuste foi de: {reaj:.2f}")
print(f"O salário após reajuste: {sal_atual + reaj:.2f}")
````

9. **[MN]** Uma iteração do método de Newton-Raphson para $f(x) = x^2 - 2$ produz $x_1 = x_0 - \dfrac{f(x_0)}{f'(x_0)}$. Leia $x_0$ e uma tolerância `tol`, calcule $x_1$ e informe se a iteração **convergiu** (isto é, se $|x_1 - x_0| < tol$) ou não.

````python
x0 = float(input("Informe o valor inicial: "))
tol = float(input("Informe a tolerância: "))

if x0 == 0:
    print("Valor inválido!")

x1 = x0 - (x0**2 - 2)/(2*x0)

if abs(x1 - x0) > tol:
    print("Não convergiu!")
else:
    print("Convergiu!")
````

13. **[MEF]** *Localização de ponto em elemento finito 1D.* Um domínio unidimensional (uma barra) foi discretizado em 3 elementos finitos lineares, com nós nas coordenadas $x_0=0$, $x_1=2$, $x_2=5$ e $x_3=10$ (metros) — o elemento 1 vai de $x_0$ a $x_1$, o elemento 2 de $x_1$ a $x_2$, e o elemento 3 de $x_2$ a $x_3$. Leia a coordenada $x$ de um ponto qualquer do domínio e, usando `if`/`elif`, determine em qual dos três elementos ele se encontra. Em seguida, calcule sua coordenada local (natural) $\xi \in [-1, 1]$ dentro do elemento identificado, usando $\xi = \dfrac{2(x - x_{esq})}{x_{dir} - x_{esq}} - 1$, em que $x_{esq}$ e $x_{dir}$ são os nós à esquerda e à direita do elemento encontrado. Essa é a primeira operação de qualquer análise por Elementos Finitos: antes de interpolar qualquer grandeza (deslocamento, temperatura, etc.) dentro de um elemento, é preciso localizar em qual elemento está o ponto de interesse.

````python
x = float(input("Digite um valor x pertencente a [0, 10]: "))

if 0 <= x <= 2:
    print("x está no elemento 1!")
    ksi = (2*(x - 0)) / (2 - 0) - 1
    print(f"O valor: {ksi:.2f}")
elif 2 <= x <= 5:
    print("x está no elemento 2!")
    ksi = (2*(x - 2)) / (5 - 2) - 1
    print(f"O valor: {ksi:.2f}")
elif 5 <= x <= 10:
    print("x está no elemento 3!")
    ksi = (2*(x - 5))/(10 - 5) - 1
    print(f"O valor: {ksi:.2f}")
````

15. **[MN]** Leia $x_0$ e calcule uma iteração de Newton-Raphson para $f(x) = x^3 - 2x - 5$ (cuja derivada é $f'(x) = 3x^2 - 2$), obtendo $x_1$. Calcule o erro $|x_1 - x_0|$ e o valor de $f(x_1)$. Classifique o resultado em três categorias: **"convergência total"** se erro $< 10^{-4}$ **e** $|f(x_1)| < 10^{-4}$; **"convergência parcial"** se apenas um dos dois critérios for satisfeito; **"sem convergência"** se nenhum for satisfeito.


````python
x0 = float(input("Digite o valor de x0: "))

fx0 = x0**3 - 2 * x0 - 5
dfx0 = 3 * x0**2 - 2

x1 = x0 - fx0/dfx0

fx1 = x1**3 - 2 * x1 - 5

er = abs(x1 - x0)
fabs = abs(fx1)

if er < 0.0001 and fabs < 0.0001:
    print("Convergencia total!")
elif er  < 0.0001 or fabs < 0.0001:
    print("Convergencia partial!")
else:
    print("Sem convergencia!")
````

18. **[MEF]** *Classificação de condição de contorno.* Uma barra de comprimento $L$ (metros), fixa na extremidade esquerda ($x=0$) e com uma força aplicada na extremidade direita ($x=L$), foi discretizada em elementos finitos cujos nós têm coordenada $x$ conhecida. Leia $L$ e a coordenada $x$ de um nó, e classifique-o (com `if`/`elif`) como: **"contorno Dirichlet"** (nó em $x=0$, onde o deslocamento é prescrito e vale zero), **"contorno Neumann"** (nó em $x=L$, onde a força é prescrita), ou **"nó interno"** (qualquer $0 < x < L$, onde deslocamento e força são incógnitas resolvidas pelo sistema). Essa classificação é o que determina, na montagem do sistema de Elementos Finitos, quais graus de liberdade entram como equação a resolver e quais entram como valor conhecido.

````python
L = float(input("Enter the bar size (L): "))
x = float(input("Enter the value of the node: "))

if x == 0:
    print("Dirichlet Boundary Condition.")
elif x == L:
    print("Neumann Boundary Condition.")
elif 0 < x < L:
    print("Internal node.")
else:
    print("Invalid node!")
````
