7. **[M]** Escreva um programa que leia o valor de um empréstimo (capital inicial $C$), a taxa de juros compostos mensal $i$ (em decimal, ex.: 0.02) e o número de meses $n$, e calcule o montante final usando $M = C(1+i)^n$.

```python
C = float(input("Digite o capital inicial: "))
i = float(input("Digite o juros compostos mensal (em decimal, ex.: 0.02): "))
n = int(input("Digite o número de meses: "))

M = C*(i+1)**n

print(f"O montante final é: {M:.2f}")
```

8. **[M]** Escreva um programa que leia a massa (kg) e a velocidade (m/s) de um objeto e calcule sua energia cinética $E = \frac{1}{2}mv^2$, imprimindo o resultado com 3 casas decimais.

```python
mass = float(input("Digite a massa do objeto(kg): "))
velocity = float(input("Digite a velocidade(m/s): "))

E = (1/2) * mass * velocity**2

print(f"A energia cinética é: {E:.3f}J")
```

9. **[MN]** O método de Newton-Raphson aproxima a raiz de uma função a partir de uma estimativa inicial $x_0$, usando $x_1 = x_0 - \dfrac{f(x_0)}{f'(x_0)}$. Escreva um programa que leia $x_0$ e calcule **uma única iteração** desse método para $f(x) = x^2 - 2$ (cuja raiz é $\sqrt{2}$), sabendo que $f'(x) = 2x$. Imprima o valor de $x_1$.

```python
x0 = float(input("Type the inicial guess for the newton method (x0 != 0): "))

x1 = x0 - (x0**2 - 2)/(2*x0) 

print(f"The root is closed to: {x1:.4f}")
```

10. **[P/M]** Escreva um programa que leia as coordenadas $(x_1, y_1)$ e $(x_2, y_2)$ de dois pontos no plano e calcule a distância euclidiana entre eles: $d = \sqrt{(x_2-x_1)^2 + (y_2-y_1)^2}$. Use `math.sqrt`.


```python
import math

x1 = float(input("Digite x1: "))
y1 = float(input("Digite y1: "))
x2 = float(input("Digite x2: "))
y2 = float(input("Digite y2: "))

d = math.sqrt((x2 - x1)**2 + (y2 - y1)**2)
print(f"A distancia entre os dois vetores são: {d:.2f}")
```

11. **[M]** O crescimento populacional exponencial é modelado por $P(t) = P_0 e^{rt}$, em que $P_0$ é a população inicial, $r$ a taxa de crescimento (decimal) e $t$ o tempo. Escreva um programa que leia $P_0$, $r$ e $t$ e calcule $P(t)$. Use `math.exp`.

```python
p0 = float(input("Digite a população incial: "))
r = float(input("Digite a taxa de crescimento: "))
t = float(input("Digite o tempo em meses: "))

P = p0*math.exp(r*t)

print(f"População após {t} meses: {P:.2f}")
```

12. **[MN]** No método da bisseção, antes de decidir qual metade do intervalo $[a, b]$ manter, calcula-se o ponto médio $m = \dfrac{a+b}{2}$ e avalia-se a função nos três pontos. Escreva um programa que leia $a$ e $b$ e imprima $m$, $f(a)$, $f(b)$ e $f(m)$ para $f(x) = x^3 - x - 2$ (sem decidir nada ainda — isso fica para quando tivermos `if`, na Aula 2).


```python
a = float(input("Digite o primeiro valor do intervalo: "))
b = float(input("Digite o segundo valor do intervalo: "))

m = (a+b)/2

fa = a**3 - a**2 - 2
fb = b**3 - b**2 - 2
fm = m**3 - m**2 - 2

print(f"Valor médio do intervalor: {m:.2f}")
print(f"f(a) = {fa:.2f} \nf(b) = {fb:.2f} \nf(m) = {fm:.2f} ")
```
