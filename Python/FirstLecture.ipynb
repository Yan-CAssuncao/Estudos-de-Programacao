2. **[P]** Escreva um programa que leia dois números inteiros e imprima a soma, a diferença, o produto, a divisão real e a divisão inteira entre eles.

```python
num1 = int(input("Digite o primeiro número: "))
num2 = int(input("Digite o segundo número: "))

print(f"A soma é: {num1 + num2}")
print(f"A diferença é: {num1 - num2}")
print(f"O produto é: {num1 * num2}")
print(f"A divisão real é: {num1 / num2}")
print(f"A divisão inteira é: {num1 // num2}")
```

3. **[P]** Escreva um programa que leia três números reais e imprima a média aritmética entre eles, com 2 casas decimais.

```python
num1 = float(input("Digite o primeiro número: "))
num2 = float(input("Digite o primeiro número: "))
num3 = float(input("Digite o primeiro número: "))

print(f"A média é: {(num1+num2+num3)/3 :.2f}")
```

4. **[M]** Escreva um programa que leia o raio de uma circunferência e imprima sua área ($A = \pi r^2$) e seu perímetro ($C = 2\pi r$). Use `math.pi` (será necessário `import math` no início do arquivo).

```python
import math

raio = float(input("Digite o raio do circulo: "))

print(f"A área é: {math.pi*raio**2:.5f}")
print(f"O perímetro é: {2*math.pi*raio:.5f}")
```
