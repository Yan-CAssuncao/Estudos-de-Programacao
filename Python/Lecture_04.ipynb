9. **[MN]** Implemente o método de Newton-Raphson para $f(x) = x^2 - 2$ armazenando **todas as iterações** em uma lista (não apenas a última). Imprima a lista completa e plote (matplotlib) o valor de $x$ ao longo das iterações.

```python
print("Enter the initial value: ")
x = float(input())
print("Enter the tolerence: ")
tol = float(input())
ite = int(input("Enter the maximum interations: "))

root_list = []
iter_list = []

i=0

while True:
    iter_list.append(i)
    root_list.append(x)
    x = x - (x**2 - 2)/(2*x)

    if i == ite:
        print("Iteration maximum!")
        break
    if (x**2 - 2) < tol:
        print(f"Root = {x}")
        break
    
    i += 1

plt.plot(iter_list, root_list, marker='o')
plt.xlabel("Iteration")
plt.ylabel("Root")
plt.title("Newton-Raphson Method for f(x)")
plt.grid()
plt.show()
```

10. **[DF]** Para a malha do exercício 5 e $f(x) = x^2$, calcule a derivada aproximada em cada ponto (progressiva no primeiro ponto, regressiva no último, central nos internos, como na Aula 3), armazenando os resultados em uma lista, e plote $x$ contra a derivada aproximada.

```python
print("Enter the initial point: ")
x = float(input())
print("Enter the number of steps: ")
n = int(input())
print("Enter the step pass h: ")
h = float(input())

knots = []
div = []

for i in range(n+1):
    if i == 0:
        y = ((x+h)**2 - x**2)/h
    if i == n:
        y = ((x)**2 - (x-h)**2)/h
    else:
        y = ((x+h)**2 - (x-h)**2)/h
    
    div.append(y)
    knots.append(x)

    x += h

plt.plot(knots, div, marker='o')
plt.xlabel("Nós")
plt.ylabel("Derivate")
plt.title("Derivate of f(x) = x²")
plt.grid()
plt.show()
```

13. **[MN]** *Bisseção com histórico de convergência.* Implemente a bisseção completa (Aula 3) para $f(x)=x^3-x-2$, armazenando em uma lista o valor de $|b-a|$ a cada iteração. Plote (com `plt.yscale("log")`) o tamanho do intervalo em função da iteração — isso evidencia visualmente que a bisseção converge **linearmente** (o erro cai por um fator aproximadamente constante a cada passo), o que será útil para comparar, mais adiante, com a velocidade de convergência de Newton-Raphson.

```python
a = float(input("Enter the a point: "))
b = float(input("Enter the b point: "))
h = float(input("Enter the tolerance h: "))
i = 0

list_values = []
interval_leg = []


while True:
    fa = a**3 - a - 2
    fb = b**3 - b - 2

    if fa * fb > 0:
        print("There not exist a root between a and b!")
        break
    c = (a+b)/2
    fc = c**3 - c - 2
    if fa*fc < 0:
        b = c
    elif fb*fc < 0:
        a = c
    
    list_values.append(c)
    interval_leg.append(i)

    i += 1
    if abs(a - b) < h or i == 100 :
        print(f"The approximate root: {c}")
        print(f"Iterations: {i}")
        break

plt.plot(interval_leg, list_values, marker='o')
plt.yscale("log")
plt.xlabel("Iteration")
plt.ylabel("Roots")
plt.title("Bissection Method")
plt.grid()
plt.show()
```

14. **[DF]** *Euler com trajetória completa, três passos.* Implemente o método de Euler para $\dfrac{dy}{dt} = -2y + t$, armazenando $t$ e $y$ de cada passo em duas listas. Repita para três valores de passo ($h=0{,}5$; $h=0{,}1$; $h=0{,}01$) e plote as três curvas na mesma figura (chame `plt.plot()` três vezes antes de `plt.show()`), com legenda (`plt.legend()`) identificando cada uma — visualizando diretamente como refinar o passo aproxima a solução numérica da solução real.

```python
import matplotlib.pyplot as plt
```

```python
# Final time:
t_final = 10.0

# Define the value of step:
h_values = [0.5, 0.1, 0.01]

for h in h_values:
        # Initial Values:
    t0 = 0.0
    y0 = 1.0
        # The list of values:
    t_list = [t0]
    y_list = [y0]

    while t_list[-1] < t_final:

        # New value of y:
        y1 = y0 + h * (-2 * y0 + t0)

        # New t0:
        t1 = t0 + h

        # Put the values in the list:
        t_list.append(t1)
        y_list.append(y1)

        # Update the value of y1:
        y0 = y1
        t0 = t1

    plt.plot(t_list, y_list, label=f"h = {h}")

plt.xlabel("t")
plt.ylabel("y")
plt.legend()
plt.grid(True)
plt.show()

print("t =", t_list)
print("y =", y_list)
```

Verifique se a soma dos valores de `F` é compatível com o equilíbrio estático de uma barra (a soma de todas as forças, incluindo reações de apoio com sinal oposto, deve ser zero) — para isso, o programa deve também pedir a reação de apoio e conferir a soma total.

```python
node = [0, 2, 5, 10]
force_node = []
sum_force = 0.0

for i in range(4):
    force_input = float(
        input(f"Enter the force of the {i+1} node: ")
    )
    force_node.append(force_input)
    sum_force += force_input

reaction = float(input("Enter the support reaction: "))

total_force = sum_force + reaction

print("Nodal force vector:", force_node)
print("Sum of nodal forces:", sum_force)
print("Total force including reaction:", total_force)


if abs(total_force) < 1e-9:
    print("Equilibrium!")
else:
    print("Not equilibrium!")
```

18. **[MEF]** *Refinamento de malha.* Para o domínio $[0, 10]$, gere (com compreensão de lista) três malhas igualmente espaçadas: com 3, 10 e 30 elementos. Para cada malha, calcule o comprimento médio de elemento e armazene os três resultados em uma lista. Plote (gráfico de linha ou de barras) o comprimento médio do elemento em função do número de elementos da malha — ilustrando o conceito de **refinamento de malha**, central tanto em Diferenças Finitas quanto em Elementos Finitos: mais elementos significam elementos menores, o que em geral (mas nem sempre, como será visto para o caso instável de Euler) melhora a precisão do método.

```python
import matplotlib.pyplot as plt

N_list = [3, 10, 30]


malhas = [[10 * i / N for i in range(N + 1)] for N in N_list]

size_mesh = [(x[-1] - x[0]) / (len(x) - 1) for x in malhas]

print(malhas)
print(size_mesh)  # [3.333..., 1.0, 0.333...]

plt.plot(N_list, size_mesh, marker="o")
plt.xlabel("Número de elementos")
plt.ylabel("Comprimento médio do elemento")
plt.title("Refinamento de malha em [0, 10]")
plt.grid(True)
plt.show()
```
