Percorrerndo listas com laços

```python
notas = []

for i in range(5):
    valor = float(input("Informe a nota: "))
    notas.append(valor)

for i in range(len(notas)):
    print(f"Nota {i+1}: {notas[i]}")
```

```python
for nota in notas:
    print(f"Nota: {nota}")
```

```python
for i, nota in enumerate(notas):
    print(f"Nota {i+1}: {nota}")
```

Exercício de Exemplo

Armazene 15 números inteiros em uma lista e, ao final, imprima todos os números lidos

```python
num = []

for i in range(5):
    n = float(input("Digite o numero inteiro: "))
    num.append(n)

for i in range(len(num)):
    print(f"Numero {i+1}: {num[i]}")

for valor in num:
    print(f"Numero: {valor}")
```

Leia 5 numeros e encontre o maior deles

```python
number = []

for i in range(5):
    val = float(input("Digite um numero: "))
    number.append(val)

maior = number[0]

for i in range(1, len(number)):
    if number[i] > maior:
        maior = number[i]

print(f"Maior = {maior}")
```

Fatiamente (slicing)

lista[2:5] - Elementos dos indices 2, 3 e 4 \
lista[:3]  - Do inicio ate o indice 2 \
lista[3:]  - Do indice ate o final \
lista[:]   - Copia da lista inteira \
lista[::-1]- A lista inteira, invertida \
lista[::2] - Elementos de indice par(passo 2) 

```python
num = [1, 2, 3, 4, 5]

print(num[1:4])
print(num[::-1])
print(num[:3])
print(num[3:])
```

Compreensão de lista (list comprehension)

```python
vet = [1, 2, 3, 4, 5]
sqrt = []

for x in vet:
    sqrt.append(x**2)
print(sqrt)
sqrt[:]
```

```python
quad = [x**2 for x in vet]
quad[:]
```

```python
pares = [x for x in vet if x % 2 == 0]
pares[:]
```

Aplique 6 iterações do método de Newton-Raphson para $f(x) = x^2 - 2$, guardando o valor de x a cada iteração, ao final plote a convergência

```python
iteracos = []
val_x = []

x = 2.0 # chute inicial

for i in range(6):
    iteracos.append(i)
    val_x.append(x)
    x = x - (x**2 - 2) / (2 * x)


plt.plot(iteracos, val_x, marker="o")
plt.ylabel("iteracao")
plt.xlabel("x")
plt.title("Convergencia do metodo de Newton-Raphson")
plt.grid()
plt.show()
```
