#### Cause ```While``` 

````python
# Read a number n and calculate the sum of all integers from 1 to n.

n = int(input("Enter the number n: "))
i = 1
sum = 0

while i <= n:
    sum += i
    i += 1

print(f"Sum: {sum}")
````

````python
# Read 10 positives numbers and print the square of each one. If the user enters a number that is not positive, 
# the program must prompt the user again until a valid value is entered.

for i in range(10):
    num = float(input("Enter a number: "))

    while num <= 0:
        print("Attention! Enter a number bigger then zero: ")
        num = float(input())

    print(f"Square: {num * num:.2f}")
````

9. **[MN]** Implement the **exact bisection** method for $f(x) = x^3 - x - 2$: read $a$, $b$, and a tolerance; repeat by reducing the interval until $|b - a|$ is less than the tolerance **or** until a maximum number of iterations (for example, 100) is reached, whichever comes first. Print the approximate root and the number of iterations used. 

````python
a = float(input("Enter the a point: "))
b = float(input("Enter the b point: "))
h = float(input("Enter the tolerance h: "))
i = 0



while True:
    fa = a**3 - a - 2
    fb = b**3 - b - 2

    if fa * fb > 0:
        print("There not exist a root between a and b!")
        break
    c = (a+b)/2
    fc = c**3 - c - 2
    if fa*fc < 0:
        b = c
    elif fb*fc < 0:
        a = c
    i += 1
    if abs(a - b) < h or i == 100 :
        print(f"The approximate root: {c}")
        print(f"Iterations: {i}")
        break
````

10. **[DF]** Implement the **full Euler** method: given $y_0$, $t_0$, a step size $h$, and a number of steps $n$, compute $y_1, y_2, \ldots, y_n$ for $\dfrac{dy}{dt} = -2y + t$, printing the values of $t$ and $y$ at each step.

````python
y0 = float(input("Enter the initial y: "))
t0 = float(input("Enter the initial time: "))
h = float(input("Enter the step size h: "))
n = int(input("Enter the number of steps: "))

i = 1


while True:
    # Euler Method
    y1 = y0 + (t0 - 2*y0)*h

    # Update time
    t1 = t0 + h


    print(f"Time: {t1:.2f} -> y = {y1:.4f}")


    if i == n:
        break

    # Prepare for the next iteration
    t0 = t1
    y0 = y1

    i += 1
````

````python
y = float(input("Enter the initial y: "))
t = float(input("Enter the initial time: "))
h = float(input("Enter the step size h: "))
n = int(input("Enter the number of steps: "))

for i in range(n):
    y = y + h*(t - 2*y)
    t = t + h

    print(f"Step {i+1}: t = {t:.2f}, y = {y:.6f}")
````

11. **[MEF]** A beam consists of $n$ spring elements connected in series, each with stiffness $k_i$. Read $n$, and in a loop, read the stiffness of each element, accumulating the equivalent stiffness of the assembly: $\dfrac{1}{k_{eq}} = \displaystyle\sum_{i=1}^{n} \dfrac{1}{k_i}$. Print $k_{eq}$.

````python
n = int(input("Enter the n spring elements: "))

keq = 0

for i in range(n):
    print("Enter the stiffness of the element: ")
    ki = float(input())
    keq += 1/ki

print(f"Stiffness equivalent: {keq:.2f}")
````

13. **[MN]** Implement the **full Newton-Raphson method, with a safety criterion**, for $f(x) = x^3 - 2x - 5$: read $x_0$ and a tolerance; repeat $x_{i+1} = x_i - f(x_i)/f'(x_i)$ until $|x_{i+1}-x_i|$ is less than the tolerance **or** until a maximum number of iterations is reached (e.g., 100)—reporting at the end whether the method converged or was terminated due to exceeding the limit.

````python
x0 = float(input("Enter the initial point: "))
h = float(input("Enter the tolerance: "))

iter_max = 100
i = 0

while True:
    fx = x0**3 - 2*x0 - 5
    df = 3*x0**2 - 2
    x1 =  x0 - fx/df

    if abs(x1 - x0) < h:
        print(f"The method converges. The root is: {x1:.4f}")
        print(f"Iteration: {i}")
        break
    elif i == iter_max:
        print("Max number of iterations!")
        break
    
    x0 = x1
    i += 1

````

14. **[DF]** *Derivatives by finite differences along a grid.* Consider a grid of $n+1$ equally spaced points, from $x_0$ to $x_n$ (step size $h$), for $f(x) = x^2$. Iterate through the points using a loop, and for each one, calculate the approximate derivative using the appropriate scheme: forward difference at the first point, backward difference at the last, and central difference at the intermediate points (combine `for` with `if`/`elif`). Print the index, the $x$ value, and the approximate derivative for each point.

````python
x = float(input("Enter the initial x: "))
n = int(input("Enter the number of steps: "))
h = float(input("Enter the step size: "))

for i in range(n+1):
    if i == 0:
        y = ((x+h)**2 - x**2)/h
    elif i == n:
        y = ((x)**2 - (x-h)**2)/h
    else:
        y = ((x+h)**2 - (x-h)**2)/h

    print(f"step: {i} --- x = {x:.4f} --- dfx = {y:.4f}")

    x += h
````
